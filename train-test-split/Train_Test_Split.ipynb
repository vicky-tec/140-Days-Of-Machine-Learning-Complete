```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

```python
tt = pd.read_csv("500hits.csv", encoding='latin1')
```

```python
tt.head(4).style.background_gradient(cmap="Dark2")
```

```python
tt.info()
```

```python
tt.isna().sum()
```

```python
tt.describe().transpose().round(2).style.background_gradient(cmap="cool") 
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = tt.drop(columns = ["PLAYER", "HOF"])
```

```python
y = tt["HOF"]
```

```python
X.head().style.background_gradient(cmap="cool")
```

```python
y.head()
```

```python
X_train, y_train, y_test, X_test = train_test_split(X, y, random_state = 42, test_size= 0.2)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
X_train.describe().round(2).style.background_gradient(cmap="Set2") \
  .format("{:.2f}")
```

```python
y_train.describe().round(2).style.background_gradient(cmap="Dark2") \
  .format("{:.2f}")
```
